# CMI Problematic Internet Use — LightGBM Model & `SequenceExtractor`

This notebook implements an end-to-end **LightGBM** pipeline for the **Child Mind Institute — Problematic Internet Use** competition, structured in the style of `baseline.ipynb` and aligned strictly with the competition rules in `Context/Child Mind Institute — Problematic Internet Use.pdf`:

- **Competition Scoring & Rules Compliance**:
  - **Primary Metric — Quadratic Weighted Kappa (QWK)** on ordinal `sii` $\in \{0, 1, 2, 3\}$ ($N = 4$ classes):
    $$\kappa = 1 - \frac{\sum_{i,j} W_{i,j} O_{i,j}}{\sum_{i,j} W_{i,j} E_{i,j}}, \quad W_{i,j} = \frac{(i - j)^2}{(N - 1)^2} = \frac{(i - j)^2}{9}$$
  - **Zero Leakage**: All `sii` and `PCIAT-*` columns (absent from `test.csv`) are stripped via `strip_leaky_columns` prior to featurization; train/holdout and CV splits occur before fitting `SequenceExtractor`, `TSAugmentor`, feature sanitization, LightGBM, and QWK thresholds.
  - **Submission Contract**: Produces `submission.csv` with exact header `id,sii`, integer predictions in `{0, 1, 2, 3}`, zero `NaN` values, and exact alignment with `sample_submission.csv` (verified by `verify_competition_submission`) under offline Code Competition constraints ($\le 9$h runtime).
- **Data Augmentation (`tsaug`)**: Time-warping, cropping, multiplicative drift, jitter, additive noise, time-shifting, temporal masking, channel dropout, and actigraphy modality dropout via `TSAugmentor` (applied strictly inside training folds).
- **Feature Extraction (`SequenceExtractor`)**: Modular multi-library extraction across `tsfel`, `tsfresh`, `tsflex`, `pycatch22` / `sktime`, `seglearn`, `statsmodels`, `scipy.signal`, `PyWavelets`, and handcrafted autocorrelation, circadian (`OpenStrap/analytics`, `cr7_kimmitch`), and sleep-architecture (`mad-lab-fau/sleep_analysis`) descriptors with leak-free missing-data imputation.
- **LightGBM Estimator (`LightGBMSequenceEstimator`)**: Supports both `LGBMRegressor` (`regression`, `huber`, `fair`, `regression_l1`, `poisson`, `tweedie`) and `LGBMClassifier` (expected-value probability projection or argmax) across `gbdt`, `dart`, and `goss` boosting types, native `NaN` split routing, leaf-wise & L1/L2/path-smoothing regularization, and leak-free internal OOF early stopping + QWK ordinal threshold calibration.
- **Separate Search Spaces**: Edit `GRID_PARAM_SPACE`, `BAYESIAN_PARAM_SPACE`, or `EVOLUTIONARY_PARAM_SPACE` in their own dedicated cells under `# Params` with zero `.update()` calls.
