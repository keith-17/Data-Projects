```python
flights_agg_df = flights_df.groupby(['flight_date', 'terminal_name']).agg(**{
    'total_flights': ('flight_number_full_icao', 'nunique'),
    'total_passengers': ('pax_quantity', 'sum'),
    'intl_flights': ('is_international', 'sum'),
    'intl_pax': ('pax_quantity', lambda x: x[flights_df.loc[x.index, 'is_international'] == 1].sum()),
    'europe_pax': ('pax_quantity', lambda x: x[flights_df.loc[x.index, 'is_europe'] == True].sum()),
    'asia_pax': ('pax_quantity', lambda x: x[flights_df.loc[x.index, 'is_asia'] == True].sum()),
    'na_pax': ('pax_quantity', lambda x: x[flights_df.loc[x.index, 'is_north_america'] == True].sum()),
    'africa_pax': ('pax_quantity', lambda x: x[flights_df.loc[x.index, 'is_africa'] == True].sum()),
}).reset_index()

flights_agg_df['flight_date'] = pd.to_datetime(flights_agg_df['flight_date'])

flights_agg_df['intl_share'] = flights_agg_df['intl_pax'] / flights_agg_df['total_passengers']
flights_agg_df['europe_share'] = flights_agg_df['europe_pax'] / flights_agg_df['total_passengers']
flights_agg_df['asia_share'] = flights_agg_df['asia_pax'] / flights_agg_df['total_passengers']
flights_agg_df['na_share'] = flights_agg_df['na_pax'] / flights_agg_df['total_passengers']
flights_agg_df['africa_share'] = flights_agg_df['africa_pax'] / flights_agg_df['total_passengers']

uk_holidays = holidays.UnitedKingdom(years=range(2017, 2026))
flights_agg_df['holiday_flag'] = flights_agg_df['flight_date'].isin(pd.to_datetime(list(uk_holidays.keys()))).astype(int)

flights_agg_df['weekofyear'] = flights_agg_df['flight_date'].dt.isocalendar().week

flights_agg_df['financial_quarter'] = flights_agg_df['flight_date'].dt.quarter

flights_agg_df.rename(columns={'flight_date': new_second_ledger_column_time_name}, inplace=True)

flights_agg_df
```

```python
group_split_list = ['weekofyear']

if use_weekly:
    print("Using weekly data for train-test split...")
    master_df_sorted = weekly_df.sample(frac=1, random_state=42).reset_index(drop=True)
else:
    # Sort by date
    master_df_sorted = master_df.drop(columns=columns_to_encode).sample(frac=1, random_state=42).reset_index(drop=True)

group_indices_series = master_df_sorted.groupby(group_split_list).ngroup()

if naive_split_bool:
    # 70/30 split
    split_idx = int(len(master_df_sorted) * 0.7)
    train_df = master_df_sorted.iloc[:split_idx]
    test_df = master_df_sorted.iloc[split_idx:]

else:
    print("Using GroupShuffleSplit for train-test split...")
    gss = GroupShuffleSplit(n_splits=1, test_size=0.25, random_state=42)
    train_idx, test_idx = next(gss.split(master_df_sorted, groups=group_indices_series))
    train_df = master_df_sorted.iloc[train_idx]
    test_df = master_df_sorted.iloc[test_idx]

print("Quarters in train:", sorted(train_df['financial_quarter'].unique()))
print("Quarters in test: ", sorted(test_df['financial_quarter'].unique()))

# Feature train target train, feature test target test
X_train = train_df.drop(columns=['occupancy'])
y_train = train_df['occupancy']
X_test = test_df.drop(columns=['occupancy'])
y_test = test_df['occupancy']
```

### XGBoost pipeline controls
`xgb_mode='filter'` preserves the existing notebook split. `xgb_mode='forecast'` builds dated historical features and reserves the latest weeks for testing.

Choose `xgb_search_mode='bayesian'` or `'grid'`. Set `xgb_horizon_weeks` to 1 or 4 for separate forecast experiments. History-window length, filtering, feature selection and XGBoost settings can be searched together.

Forecasts use observations through the end of the origin week. The fitted model stays fixed during the final holdout; its input history updates at each weekly origin. Forecast scores are reported separately from the older random-split leaderboard.

All controls are at the top of the next cell. Helper classes live in `xgboost_sequence.py` beside this notebook. Bayesian search requires `scikit-optimize` in the active kernel; install with `%pip install scikit-optimize` if needed. XGBoost early stopping is deliberately disabled because it needs a separate validation set inside each training fold.


```python
# XGBoost: sequence/filter pipeline, optional Bayesian or grid search.
# Run this cell after the notebook's existing train/test preparation.

# ============================================================
# 1. CONFIGURATION
# ============================================================
xgb_mode = 'filter'                 # 'filter' = existing split; 'forecast' = dated history
xgb_search_mode = 'bayesian'        # 'bayesian' or 'grid'
xgb_horizon_weeks = 1               # Forecast mode only: run 1 and 4 separately
xgb_history_windows = (4, 8)        # Searchable lookback lengths; same rows for every trial
xgb_forecast_year = 2023            # None keeps all available target years
xgb_history_columns = ('occupancy', 'total_flights', 'total_passengers')
xgb_holdout_fraction = 0.25         # Final chronological weeks in forecast mode
xgb_cv_splits = 3
xgb_cv_test_size = 0.15             # Fraction of distinct training weeks per CV fold
xgb_bayes_iterations = 50
xgb_search_jobs = 4                 # Parallel CV fits; use 1 for debugging
xgb_model_jobs = 1                  # Avoid parallelising both layers across every CPU
xgb_search_verbose = 3            # 0 = quiet; higher values show CV progress
xgb_model_verbose = 3              # XGBoost: 0 silent, 1 warnings, 2 info, 3 debug
xgb_sequence_verbose = 2           # 1 reports training rows/features retained per fit
xgb_show_results = True
xgb_show_all_parameters = False
xgb_return_train_score = True
xgb_error_score = 'raise'           # Fail visibly if a candidate is invalid
xgb_scale_features = False         # Optional StandardScaler, fitted within each CV fold
xgb_clip_nonnegative = True         # Same clipping used during CV and holdout prediction
xgb_save_results = False
xgb_random_state = 42

if run_xgboost:
    import importlib
    from pathlib import Path
    import numpy as np
    import pandas as pd
    from IPython.display import display
    from sklearn.pipeline import Pipeline
    from sklearn.model_selection import GridSearchCV, ParameterGrid
    from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error
    from xgboost import XGBRegressor
    import xgboost_sequence
    importlib.reload(xgboost_sequence)
    from xgboost_sequence import (
        CavuSequenceExtractor, SequenceFilteredRegressor,
        build_forecast_bank, chronological_week_splits,
    )

    if xgb_mode not in {'filter', 'forecast'}:
        raise ValueError("xgb_mode must be 'filter' or 'forecast'.")
    if xgb_search_mode not in {'bayesian', 'grid'}:
        raise ValueError("xgb_search_mode must be 'bayesian' or 'grid'.")
    if xgb_horizon_weeks < 1 or int(xgb_horizon_weeks) != xgb_horizon_weeks:
        raise ValueError('xgb_horizon_weeks must be a positive integer.')
    if not xgb_history_windows or any(int(w) != w or w < 1 for w in xgb_history_windows):
        raise ValueError('xgb_history_windows must contain positive integers.')

    # ========================================================
    # 2. DATA AND CROSS-VALIDATION
    # ========================================================
    if xgb_mode == 'filter':
        # Preserve the exact rows/order used by the other notebook models.
        xgb_X_train, xgb_X_test = X_train.copy(), X_test.copy()
        xgb_y_train, xgb_y_test = y_train.copy(), y_test.copy()
        xgb_cv_groups = train_df.loc[xgb_X_train.index, 'weekofyear']
        xgb_cv = general_utils.make_group_cv_splitter(
            xgb_cv_groups, n_splits=xgb_cv_splits,
            test_size=xgb_cv_test_size, random_state=xgb_random_state,
        )
        xgb_test_context = xgb_X_test.copy()
        xgb_evaluation = 'Existing grouped/random week holdout; no fixed forecast horizon'
    else:
        if not use_weekly:
            raise ValueError('Forecast mode requires use_weekly=True and weekly_meta.')
        xgb_bank, xgb_targets, xgb_meta = build_forecast_bank(
            weekly_df, weekly_meta,
            horizon_weeks=xgb_horizon_weeks,
            max_history_weeks=max(xgb_history_windows),
            history_columns=xgb_history_columns,
        )
        if xgb_forecast_year is not None:
            xgb_keep = xgb_meta['week_start'].dt.year.eq(xgb_forecast_year)
            xgb_bank = xgb_bank.loc[xgb_keep]
            xgb_targets = xgb_targets.loc[xgb_keep]
            xgb_meta = xgb_meta.loc[xgb_keep]
        xgb_weeks = pd.DatetimeIndex(sorted(xgb_meta['week_start'].unique()))
        if not 0 < xgb_holdout_fraction < 1:
            raise ValueError('xgb_holdout_fraction must be between zero and one.')
        xgb_n_test_weeks = max(1, int(np.ceil(len(xgb_weeks) * xgb_holdout_fraction)))
        if xgb_n_test_weeks >= len(xgb_weeks):
            raise ValueError('Insufficient dated history for a train/holdout split.')
        xgb_first_test = xgb_weeks[-xgb_n_test_weeks]
        # A four-week target cannot train on labels from the intervening three weeks.
        xgb_last_known_week = xgb_first_test - pd.Timedelta(weeks=xgb_horizon_weeks)
        xgb_train_mask = xgb_meta['week_start'].le(xgb_last_known_week)
        xgb_test_mask = xgb_meta['week_start'].ge(xgb_first_test)
        xgb_X_train, xgb_X_test = xgb_bank.loc[xgb_train_mask], xgb_bank.loc[xgb_test_mask]
        xgb_y_train, xgb_y_test = xgb_targets.loc[xgb_train_mask], xgb_targets.loc[xgb_test_mask]
        xgb_train_meta, xgb_test_meta = xgb_meta.loc[xgb_train_mask], xgb_meta.loc[xgb_test_mask]
        xgb_cv = chronological_week_splits(
            xgb_train_meta['week_start'], horizon_weeks=xgb_horizon_weeks,
            n_splits=xgb_cv_splits, test_size=xgb_cv_test_size,
        )
        xgb_cv_groups = None
        # Context is for reporting only. These contemporaneous columns never enter X.
        xgb_test_context = weekly_df.loc[xgb_X_test.index].drop(columns='occupancy').copy()
        xgb_test_context['date'] = xgb_test_meta['week_start']
        xgb_test_context['forecast_origin_week'] = xgb_test_meta['origin_week']
        xgb_test_context['horizon_weeks'] = xgb_horizon_weeks
        xgb_evaluation = (
            f'Chronological {xgb_horizon_weeks}-week forecasting; '
            'model fitted once before holdout, history updated at each forecast origin'
        )

    if not xgb_X_train.index.equals(xgb_y_train.index) or not xgb_X_test.index.equals(xgb_y_test.index):
        raise ValueError('X and y must have identical indices and row order.')
    if xgb_X_train.empty or xgb_X_test.empty:
        raise ValueError('Training and holdout data must both contain observations.')
    if xgb_X_train.index.intersection(xgb_X_test.index).size:
        raise ValueError('Training and holdout row indices overlap.')

    # ========================================================
    # 3. PIPELINE: EXTRACTOR / FILTER + XGBOOST
    # ========================================================
    # The wrapper filters X AND y inside fit, never inside predict.
    # Imputation, filtering, selection and optional scaling learn from each CV fit only.
    xgb_extractor_params = dict(
        mode=xgb_mode,
        horizon_weeks=xgb_horizon_weeks,
        history_window=max(xgb_history_windows),
        include_lags=True,
        include_rolling=True,
        filter_method='none',
        filter_threshold=1.5,
        max_outlier_features=0,
        select_features=False,
        n_features=12,
    )
    # Add any other supported XGBRegressor constructor parameter to this dictionary.
    xgb_model_params = dict(
        objective='reg:squarederror', tree_method='hist',
        n_estimators=300, max_depth=4, learning_rate=0.05,
        min_child_weight=1, gamma=0.0,
        subsample=0.8, colsample_bytree=0.8,
        colsample_bylevel=1.0, colsample_bynode=1.0,
        reg_alpha=0.0, reg_lambda=1.0, max_bin=256,
        random_state=xgb_random_state,
        n_jobs=xgb_model_jobs, verbosity=xgb_model_verbose,
    )
    xgb_model_pipeline = Pipeline([
        ('estimator', SequenceFilteredRegressor(
            extractor=CavuSequenceExtractor(**xgb_extractor_params),
            estimator=XGBRegressor(**xgb_model_params),
            scale_features=xgb_scale_features,
            clip_nonnegative=xgb_clip_nonnegative,
            verbose=xgb_sequence_verbose,
        )),
    ])
    xgb_seq_prefix = 'estimator__extractor__'
    xgb_model_prefix = 'estimator__estimator__'

    # ========================================================
    # 4. GRID AND BAYESIAN PARAMETER SPACES
    # ========================================================
    # Keep grid searches small: candidate counts multiply across every list.
    xgb_grid_param_space = {
        xgb_seq_prefix + 'filter_method': ['none', 'iqr', 'zscore'],
        xgb_model_prefix + 'n_estimators': [37,100, 200, 300, 500],
        xgb_model_prefix + 'max_depth': [6, 37, 50, 100],
        xgb_model_prefix + 'learning_rate': [0.001, 0.01, 0.05, 0.1, 0.15],
        xgb_model_prefix + 'subsample': [0.6, 0.8, 1.0],
        xgb_model_prefix + 'colsample_bytree': [0.6, 0.8, 1.0],
        # Optional examples; uncomment the parameters you actually want to search:
        xgb_seq_prefix + 'filter_threshold': [1.5, 3.0],
        xgb_seq_prefix + 'max_outlier_features': [0, 5, 10, 20],
        xgb_seq_prefix + 'select_features': [False, True],
        xgb_seq_prefix + 'n_features': [8, 12, 16, 20, 24],
        xgb_model_prefix + 'min_child_weight': [1, 5, 10, 20],
        xgb_model_prefix + 'gamma': [0.0, 0.5, 1.0],
        xgb_model_prefix + 'reg_alpha': [0.0, 0.5, 1.0],
        xgb_model_prefix + 'reg_lambda': [1.0, 5.0, 10.0],
    }
    if xgb_mode == 'forecast':
        xgb_grid_param_space[xgb_seq_prefix + 'history_window'] = list(xgb_history_windows)
        # include_lags/include_rolling are also searchable extractor parameters.
        # Horizon is fixed per experiment because it defines the business task.

    xgb_bayesian_param_space = None
    try:
        from skopt import BayesSearchCV
        from skopt.space import Categorical, Integer, Real
    except ImportError:
        if xgb_search_mode == 'bayesian':
            raise ImportError(
                "Bayesian search needs scikit-optimize in this notebook kernel. "
                "Run %pip install scikit-optimize, or set xgb_search_mode='grid'."
            ) from None
    else:
        xgb_bayesian_param_space = {
            xgb_seq_prefix + 'filter_method': Categorical(['none', 'iqr', 'zscore']),
            xgb_seq_prefix + 'filter_threshold': Real(1.5, 4.0),
            xgb_seq_prefix + 'max_outlier_features': Integer(0, 3),
            xgb_seq_prefix + 'select_features': Categorical([False, True]),
            xgb_seq_prefix + 'n_features': Integer(4, max(5, min(30, xgb_X_train.shape[1]))),
            xgb_model_prefix + 'n_estimators': Integer(100, 500),
            xgb_model_prefix + 'max_depth': Integer(2, 9),
            xgb_model_prefix + 'learning_rate': Real(0.01, 0.2, prior='log-uniform'),
            xgb_model_prefix + 'min_child_weight': Real(1.0, 12.0, prior='log-uniform'),
            xgb_model_prefix + 'gamma': Real(0.0, 3.0),
            xgb_model_prefix + 'subsample': Real(0.6, 1.0),
            xgb_model_prefix + 'colsample_bytree': Real(0.6, 1.0),
            xgb_model_prefix + 'reg_alpha': Real(1e-6, 5.0, prior='log-uniform'),
            xgb_model_prefix + 'reg_lambda': Real(0.1, 10.0, prior='log-uniform'),
        }
        if xgb_mode == 'forecast':
            xgb_bayesian_param_space[xgb_seq_prefix + 'history_window'] = Categorical(list(xgb_history_windows))

    # Validate names before launching an expensive search.
    xgb_available_params = xgb_model_pipeline.get_params(deep=True)
    for xgb_space in (xgb_grid_param_space, xgb_bayesian_param_space):
        if xgb_space is not None:
            xgb_unknown = sorted(set(xgb_space) - set(xgb_available_params))
            if xgb_unknown:
                raise ValueError(f'Unknown pipeline parameters: {xgb_unknown}')
            if xgb_seq_prefix + 'horizon_weeks' in xgb_space:
                raise ValueError('Evaluate horizons separately; do not tune the target horizon.')
    if xgb_show_all_parameters:
        display(pd.Series(xgb_available_params, name='Current value').to_frame())

    # ========================================================
    # 5. SEARCH AND REFIT ON TRAINING DATA
    # ========================================================
    xgb_search_options = dict(
        estimator=xgb_model_pipeline,
        scoring='neg_root_mean_squared_error', cv=xgb_cv,
        n_jobs=xgb_search_jobs, pre_dispatch=xgb_search_jobs if xgb_search_jobs > 0 else '2*n_jobs',
        verbose=xgb_search_verbose, error_score=xgb_error_score,
        return_train_score=xgb_return_train_score, refit=True,
    )
    if xgb_search_mode == 'grid':
        xgb_search = GridSearchCV(param_grid=xgb_grid_param_space, **xgb_search_options)
        xgb_candidates = len(ParameterGrid(xgb_grid_param_space))
    else:
        xgb_search = BayesSearchCV(
            search_spaces=xgb_bayesian_param_space,
            n_iter=xgb_bayes_iterations, n_points=1,
            random_state=xgb_random_state, **xgb_search_options,
        )
        xgb_candidates = xgb_bayes_iterations
    if xgb_show_results:
        print(xgb_evaluation)
        print(f'{xgb_search_mode.title()} search: {xgb_candidates} candidates; {len(xgb_X_train):,} training rows; {len(xgb_X_test):,} holdout rows')
    xgb_fit_options = {} if xgb_cv_groups is None else {'groups': xgb_cv_groups}
    xgb_search.fit(xgb_X_train, xgb_y_train, **xgb_fit_options)
    xgb_pipeline = xgb_search             # Existing notebook compatibility
    xgb_best_pipeline = xgb_search.best_estimator_
    xgb_best_model = xgb_best_pipeline.named_steps['estimator'].estimator_
    xgb_best_extractor = xgb_best_pipeline.named_steps['estimator'].extractor_
    y_pred_xgb_array = xgb_best_pipeline.predict(xgb_X_test)
    xgb_cv_results_df = pd.DataFrame(xgb_search.cv_results_).sort_values('rank_test_score')

    # ========================================================
    # 6. HOLDOUT LEDGER, METRICS AND OPTIONAL EXPORTS
    # ========================================================
    xgb_ledger_df = xgb_test_context.copy()
    xgb_ledger_df['ground_truth'] = xgb_y_test
    xgb_ledger_df['model_prediction'] = y_pred_xgb_array
    xgb_ledger_df['evaluation'] = xgb_evaluation
    xgb_ledger_df['raw_difference'] = xgb_ledger_df['model_prediction'] - xgb_ledger_df['ground_truth']
    xgb_ledger_df['mae'] = xgb_ledger_df['raw_difference'].abs()
    xgb_ledger_df['mse'] = xgb_ledger_df['raw_difference'].pow(2)
    # Legacy row columns: row RMSE equals absolute error; group metrics are recomputed below.
    xgb_ledger_df['rmse'] = np.sqrt(xgb_ledger_df['mse'])
    xgb_variance = np.var(xgb_y_test.to_numpy(dtype=float))
    xgb_ledger_df['r2'] = 1 - xgb_ledger_df['mse'] / xgb_variance if xgb_variance > 0 else np.nan
    xgb_denominator = xgb_ledger_df['ground_truth'].replace(0, np.nan)
    xgb_ledger_df['perc_difference'] = 100 * xgb_ledger_df['raw_difference'] / xgb_denominator
    # Zero actual: a zero prediction is correct; any nonzero miss is extreme.
    xgb_zero_actual = xgb_ledger_df['ground_truth'].eq(0)
    xgb_ledger_df.loc[xgb_zero_actual, 'perc_difference'] = np.where(
        xgb_ledger_df.loc[xgb_zero_actual, 'raw_difference'].eq(0), 0,
        np.sign(xgb_ledger_df.loc[xgb_zero_actual, 'raw_difference']) * np.inf,
    )
    xgb_pct = xgb_ledger_df['perc_difference'].to_numpy()
    xgb_labels = ['Extreme Underforecast', 'Underforecast', 'Correct', 'Overforecast', 'Extreme Overforecast', 'No output']
    xgb_ledger_df['accuracy_bin'] = pd.Categorical(np.select(
        [xgb_pct < -15, xgb_pct < -5, xgb_pct <= 5, xgb_pct <= 15, xgb_pct > 15],
        xgb_labels[:5], default='No output',
    ), categories=xgb_labels, ordered=True)
    r2_xgb = r2_score(xgb_y_test, y_pred_xgb_array)
    mse_xgb = mean_squared_error(xgb_y_test, y_pred_xgb_array)
    rmse_xgb = np.sqrt(mse_xgb)
    mae_xgb = mean_absolute_error(xgb_y_test, y_pred_xgb_array)
    xgb_metrics_df = pd.DataFrame([{
        'Model': 'XGBoost', 'Evaluation': xgb_evaluation,
        'R2 Score': r2_xgb, 'RMSE': rmse_xgb, 'MAE': mae_xgb, 'MSE': mse_xgb,
    }])
    xgb_feature_importances_df = pd.DataFrame({
        'Feature': xgb_best_extractor.get_feature_names_out(),
        'Importance': xgb_best_model.feature_importances_,
    }).sort_values('Importance', ascending=False)
    if xgb_show_results:
        display(xgb_metrics_df)
        print('Best parameters:', xgb_search.best_params_)
        print(f'Best mean CV RMSE: {-xgb_search.best_score_:.2f}')
        print(f'Refit retained {xgb_best_extractor.n_samples_retained_:,}/{xgb_best_extractor.n_samples_in_:,} training rows and {len(xgb_best_extractor.get_feature_names_out())} features')
        if xgb_mode == 'forecast':
            print('Forecast results have their own dates/targets. The earlier random-split model leaderboard excludes this experiment.')
    if xgb_save_results:
        xgb_output_directory = Path(reports_directory) / 'xgboost_search'
        xgb_output_directory.mkdir(parents=True, exist_ok=True)
        xgb_tag = f'{xgb_mode}_{xgb_horizon_weeks}w_{xgb_search_mode}'
        xgb_cv_results_df.to_csv(xgb_output_directory / f'{xgb_tag}_cv.csv', index=False)
        xgb_ledger_df.to_csv(xgb_output_directory / f'{xgb_tag}_holdout.csv', index_label='source_row')
        xgb_metrics_df.to_csv(xgb_output_directory / f'{xgb_tag}_metrics.csv', index=False)
else:
    print('Skipping XGBoost.')
```

```python
if run_xgboost:
    def summarise_xgb_group(group):
        actual = group['ground_truth'].to_numpy(dtype=float)
        predicted = group['model_prediction'].to_numpy(dtype=float)
        return {
            'mae': mean_absolute_error(actual, predicted),
            'mse': mean_squared_error(actual, predicted),
            'rmse': np.sqrt(mean_squared_error(actual, predicted)),
            'r2': r2_score(actual, predicted) if len(actual) > 1 and np.var(actual) > 0 else np.nan,
            'Records Used': len(group),
        }

    xgb_summary_rows = []
    for group_columns in [
        ['terminal_num'],
        ['terminal_num', 'carpark_name_num'],
        ['financial_quarter', 'terminal_num', 'carpark_name_num'],
    ]:
        for keys, group in xgb_ledger_df.groupby(group_columns, observed=True):
            if not isinstance(keys, tuple):
                keys = (keys,)
            row = dict(zip(group_columns, keys))
            row.update(summarise_xgb_group(group))
            row['Evaluation'] = xgb_evaluation
            xgb_summary_rows.append(row)
    xgb_result_df = pd.DataFrame(xgb_summary_rows)
    xgb_result_df['terminal_num'] += 1
    xgb_result_df['carpark_name_num'] = xgb_result_df['carpark_name_num'].add(1).astype(object)
    xgb_result_df = xgb_result_df.fillna({
        'financial_quarter': 'Year Round', 'carpark_name_num': 'All Car Parks',
    }).rename(columns={
        'terminal_num': 'Terminal Number', 'carpark_name_num': 'Carpark Name',
        'financial_quarter': 'Financial Quarter',
    })
    display(xgb_result_df)
else:
    print('Skipping XGBoost result summaries.')
```

```python
if run_neural_network:
    from tensorflow.keras.callbacks import EarlyStopping
    from tensorflow.keras.layers import Dense, Dropout
    from tensorflow.keras.models import Sequential
    from tensorflow.keras.optimizers import Adam
    nn_features = [col for col in X_train.columns if col != 'occupancy']
    scaler = StandardScaler()
    X_train_nn = scaler.fit_transform(X_train[nn_features].fillna(0))
    X_test_nn = scaler.transform(X_test[nn_features].fillna(0))

    model = Sequential([
        Dense(128, activation='relu', input_shape=(X_train_nn.shape[1],)),
        Dropout(0.2),
        Dense(64, activation='relu'),
        Dropout(0.1),
        Dense(1)
    ])
    model.compile(optimizer=Adam(learning_rate=0.001), loss='mse', metrics=['mae'])

    early_stop = EarlyStopping(monitor='val_loss', patience=10, restore_best_weights=True)
    history = model.fit(
        X_train_nn, y_train.values,
        validation_split=0.2,
        epochs=100,
        batch_size=32,
        callbacks=[early_stop],
        verbose=0
    )

    y_pred_nn = model.predict(X_test_nn).flatten()
    r2_nn = r2_score(y_test, y_pred_nn)
    rmse_nn = np.sqrt(mean_squared_error(y_test, y_pred_nn))
    mae_nn = mean_absolute_error(y_test, y_pred_nn)
    mse_nn = mean_squared_error(y_test, y_pred_nn)

    print(f"Neural Network R2 Score: {r2_nn:.4f}")
    print(f"Neural Network RMSE: {rmse_nn:.2f}")
    print(f"Neural Network MAE: {mae_nn:.2f}")
    print(f"Neural Network MSE: {mse_nn:.2f}")

    plt.figure(figsize=(8, 5))
    plt.plot(history.history['loss'], label='Training Loss')
    plt.plot(history.history['val_loss'], label='Validation Loss')
    plt.xlabel('Epoch')
    plt.ylabel('MSE Loss')
    plt.title('Neural Network Training & Validation Loss')
    plt.legend()
    plt.grid(True)
    plt.tight_layout()
    plt.show()
else:
    print("Skipping Neural Network.")
```

```python
import seaborn as sns
from matplotlib.ticker import FuncFormatter


def plot_metric_scores(df, top_model='XGBoost', suffix=''):
    """One chart per metric, with the top model first and every score labelled."""
    df = df.copy()
    df.loc[df['Model'] == 'Linear Regression', 'Model'] = '(Baseline) Linear Regression'
    higher_is_better = {'R2 Score': True, 'RMSE': False, 'MAE': False, 'MSE': False}
    highlight = '#e07b00'
    others = ['#1f4e79', '#2a9d8f', '#8e44ad', '#c0392b', '#6c8ead', '#7f8c8d', '#b8860b']

    for metric, higher in higher_is_better.items():
        rest = df[df['Model'] != top_model].sort_values(metric, ascending=not higher)
        top = df[df['Model'] == top_model]
        plot_df = pd.concat([top, rest])
        order = plot_df['Model'].tolist()
        other_models = [m for m in order if m != top_model]
        palette = {m: others[i % len(others)] for i, m in enumerate(other_models)}
        palette[top_model] = highlight

        fig, ax = plt.subplots(figsize=(10, max(2.5, 0.8 * len(order) + 1.5)))
        sns.set_theme(style='white')
        sns.barplot(data=plot_df, y='Model', x=metric, order=order, hue='Model',
                    palette=palette, legend=False, ax=ax)
        for container in ax.containers:
            ax.bar_label(container, fmt='{:,.3f}' if metric == 'R2 Score' else '{:,.1f}',
                         padding=6, fontsize=11, fontweight='bold', color='#333333')
        ax.margins(x=0.15)
        ax.set_title(f"{metric} - {'higher' if higher else 'lower'} is better{suffix}",
                     fontsize=14, fontweight='bold', loc='left')
        ax.set_xlabel(f'{metric} Score' if not metric.endswith('Score') else metric, fontsize=11, fontweight='bold')
        ax.set_ylabel('Model', fontsize=11, fontweight='bold')
        ax.xaxis.set_major_formatter(FuncFormatter(lambda v, _: f'{v:,.2f}' if metric == 'R2 Score' else f'{v:,.0f}'))
        ax.xaxis.grid(True, color='grey', alpha=0.25, linewidth=0.8)
        ax.set_axisbelow(True)
        sns.despine(ax=ax, left=True)
        plt.tight_layout()
        plt.show()

if metrics_df.empty:
    print("Skipping model metric plots; all model switches are disabled.")
else:
    metrics_df_no_arima = metrics_df[metrics_df['Model'] != 'ARIMA']
    if not metrics_df_no_arima.empty:
        plot_metric_scores(metrics_df_no_arima, suffix=' (excluding ARIMA)')

    metrics_arima = metrics_df[metrics_df['Model'] == 'ARIMA']
    if not metrics_arima.empty:
        plot_metric_scores(metrics_arima, top_model='ARIMA', suffix=' (ARIMA)')
```

```python
if run_random_forest:
    import matplotlib.pyplot as plt
    import numpy as np
    import pandas as pd
    from matplotlib.patches import Patch
    from matplotlib.ticker import StrMethodFormatter


    def neighbouring_week_interpolation(train_df, test_df, group_cols, time_col, target_col):
        """Interpolate each test row from training observations within its group."""
        required_columns = set(group_cols) | {time_col, target_col}
        missing_columns = required_columns - set(train_df.columns)
        missing_columns |= required_columns - set(test_df.columns)
        if missing_columns:
            raise KeyError(f"Missing required columns: {sorted(missing_columns)}")

        train_by_group = {}
        for group_key, group in train_df.groupby(group_cols, dropna=False, sort=False):
            if not isinstance(group_key, tuple):
                group_key = (group_key,)

            points = group[[time_col, target_col]].apply(pd.to_numeric, errors='coerce').dropna()
            points = points[np.isfinite(points[time_col]) & np.isfinite(points[target_col])]
            weekly_points = points.groupby(time_col)[target_col].mean().sort_index()
            if not weekly_points.empty:
                train_by_group[group_key] = weekly_points

        predictions = pd.Series(np.nan, index=test_df.index, dtype=float, name='interpolated_prediction')
        for group_key, group in test_df.groupby(group_cols, dropna=False, sort=False):
            if not isinstance(group_key, tuple):
                group_key = (group_key,)

            if group_key not in train_by_group:
                raise ValueError(f"No training observations found for test group {group_key}.")

            weekly_points = train_by_group[group_key]
            test_weeks = pd.to_numeric(group[time_col], errors='coerce').to_numpy(dtype=float)
            if not np.isfinite(test_weeks).all():
                raise ValueError(f"Test data contains invalid values in {time_col!r}.")

            predictions.loc[group.index] = np.interp(
                test_weeks,
                weekly_points.index.to_numpy(dtype=float),
                weekly_points.to_numpy(dtype=float),
            )

        result = test_df[group_cols + [time_col]].copy()
        result['interpolated_prediction'] = predictions
        return result


    def simulate_revenue(actual_occupancy, predicted_occupancy, capacity, base_price, elasticity=-1.2):
        """Return realized revenue per row under the occupancy-based price rules."""
        actual_values, predicted_values, capacity_values = np.broadcast_arrays(
            np.asarray(actual_occupancy, dtype=float),
            np.asarray(predicted_occupancy, dtype=float),
            np.asarray(capacity, dtype=float),
        )
        if np.any(capacity_values <= 0):
            raise ValueError("capacity must be greater than zero.")

        predicted_rate = predicted_values / capacity_values
        multiplier = np.where(
            predicted_rate > 0.85,
            1.30,
            np.where(predicted_rate < 0.50, 0.80, 1.00),
        )
        demand = actual_values * multiplier ** elasticity
        cars_sold = np.minimum(demand, capacity_values)
        revenue = cars_sold * base_price * multiplier

        if isinstance(actual_occupancy, pd.Series):
            return pd.Series(revenue, index=actual_occupancy.index, name='revenue')
        return revenue


    def calculate_perfect_forecast_revenue(actual_occupancy, capacity, base_price, elasticity=-1.2):
        """Return the best revenue across all available price multipliers per row."""
        actual_values, capacity_values = np.broadcast_arrays(
            np.asarray(actual_occupancy, dtype=float),
            np.asarray(capacity, dtype=float),
        )
        if np.any(capacity_values <= 0):
            raise ValueError("capacity must be greater than zero.")

        multipliers = np.array([1.30, 1.00, 0.80])
        demand = actual_values[..., np.newaxis] * multipliers ** elasticity
        cars_sold = np.minimum(demand, capacity_values[..., np.newaxis])
        candidate_revenue = cars_sold * base_price * multipliers
        revenue = candidate_revenue.max(axis=-1)

        if isinstance(actual_occupancy, pd.Series):
            return pd.Series(revenue, index=actual_occupancy.index, name='revenue')
        return revenue


    # The weekly frame stores these identifiers as numeric codes after aggregation.
    if {'terminal', 'carpark_name'}.issubset(train_df.columns):
        group_cols = ['terminal', 'carpark_name']
    else:
        group_cols = ['terminal_num', 'carpark_name_num']

    time_col = 'weekofyear'
    target_col = 'occupancy'
    base_price = 20.0
    train_mean = train_df[target_col].mean()

    month_means = train_df.groupby(group_cols + ['month'], dropna=False)[target_col].mean()
    carpark_means = train_df.groupby(group_cols, dropna=False)[target_col].mean()
    baseline_month_keys = pd.MultiIndex.from_frame(test_df[group_cols + ['month']])
    baseline_carpark_keys = pd.MultiIndex.from_frame(test_df[group_cols])
    baseline_predictions = pd.Series(
        month_means.reindex(baseline_month_keys).to_numpy(),
        index=test_df.index,
    ).fillna(
        pd.Series(carpark_means.reindex(baseline_carpark_keys).to_numpy(), index=test_df.index)
    ).fillna(train_mean)

    interpolation_predictions_df = neighbouring_week_interpolation(
        train_df=train_df,
        test_df=test_df,
        group_cols=group_cols,
        time_col=time_col,
        target_col=target_col,
    )

    rf_predictions = pd.Series(
        rf_corr.predict(X_test_corr),
        index=X_test_corr.index,
        name='rf_prediction',
    ).reindex(test_df.index).fillna(train_mean)

    revenue_eval_df = test_df[group_cols + [time_col, target_col, 'financial_quarter']].copy()
    revenue_eval_df['capacity'] = master_df.groupby(group_cols, dropna=False)[target_col].max().mul(7).reindex(
        pd.MultiIndex.from_frame(revenue_eval_df[group_cols])
    ).to_numpy()
    revenue_eval_df['baseline_prediction'] = baseline_predictions.reindex(revenue_eval_df.index).fillna(train_mean)
    revenue_eval_df['interpolation_prediction'] = interpolation_predictions_df[
        'interpolated_prediction'
    ].reindex(revenue_eval_df.index).fillna(train_mean)
    revenue_eval_df['rf_prediction'] = rf_predictions

    revenue_eval_df['baseline_revenue'] = simulate_revenue(
        revenue_eval_df[target_col],
        revenue_eval_df['baseline_prediction'],
        revenue_eval_df['capacity'],
        base_price,
    )
    revenue_eval_df['interpolation_revenue'] = simulate_revenue(
        revenue_eval_df[target_col],
        revenue_eval_df['interpolation_prediction'],
        revenue_eval_df['capacity'],
        base_price,
    )
    revenue_eval_df['rf_revenue'] = simulate_revenue(
        revenue_eval_df[target_col],
        revenue_eval_df['rf_prediction'],
        revenue_eval_df['capacity'],
        base_price,
    )
    revenue_eval_df['perfect_forecast_revenue'] = calculate_perfect_forecast_revenue(
        revenue_eval_df[target_col],
        revenue_eval_df['capacity'],
        base_price,
    )

    weekly_revenue_df = revenue_eval_df.groupby(time_col)[
        ['baseline_revenue', 'rf_revenue', 'perfect_forecast_revenue', 'interpolation_revenue']
    ].sum().sort_index()
    all_weeks = pd.RangeIndex(1, int(master_df[time_col].max()) + 1, name=time_col)
    weekly_revenue_df = weekly_revenue_df.reindex(all_weeks, fill_value=0)
    weekly_revenue_df.columns = [
        'Monthly Carpark Average',
        'Random Forest Model',
        'Perfect Forecast',
        'Interpolation Baseline',
    ]
    cumulative_revenue_df = weekly_revenue_df.cumsum()

    baseline_total = cumulative_revenue_df['Monthly Carpark Average'].iloc[-1]
    rf_total = cumulative_revenue_df['Random Forest Model'].iloc[-1]
    perfect_forecast_total = cumulative_revenue_df['Perfect Forecast'].iloc[-1]
    interpolation_total = cumulative_revenue_df['Interpolation Baseline'].iloc[-1]
    rf_uplift = rf_total - baseline_total
    rf_uplift_pct = (rf_uplift / baseline_total * 100) if baseline_total else np.nan
    achievable_gain = perfect_forecast_total - baseline_total
    share_of_achievable_gain = (rf_uplift / achievable_gain) if achievable_gain else np.nan

    summary_table = pd.DataFrame(
        {
            'Total Cumulative Revenue (£)': [baseline_total, rf_total, perfect_forecast_total, interpolation_total, np.nan],
            'Revenue Uplift (£)': [np.nan, rf_uplift, np.nan, np.nan, rf_uplift],
            'Revenue Uplift (%)': [np.nan, rf_uplift_pct, np.nan, np.nan, rf_uplift_pct],
        },
        index=[
            'Monthly Carpark Average',
            'Random Forest Model',
            'Perfect Forecast',
            'Interpolation Baseline',
            'RF Uplift vs Baseline',
        ],
    )
    print(summary_table.to_string(formatters={
        'Total Cumulative Revenue (£)': lambda value: f"£{value:,.2f}" if pd.notna(value) else 'N/A',
        'Revenue Uplift (£)': lambda value: f"£{value:,.2f}" if pd.notna(value) else 'N/A',
        'Revenue Uplift (%)': lambda value: f"{value:,.2f}%" if pd.notna(value) else 'N/A',
    }))
    print(f"Share of Achievable Gain Captured: {share_of_achievable_gain:.2%}")
    print(f"Perfect Forecast total >= RF total: {perfect_forecast_total >= rf_total}")
    print(f"Perfect Forecast total >= baseline total: {perfect_forecast_total >= baseline_total}")

    actual_values = revenue_eval_df[target_col].to_numpy(dtype=float)
    capacity_values = revenue_eval_df['capacity'].to_numpy(dtype=float)
    price_tiers = np.array([1.30, 1.00, 0.80])
    baseline_rate = revenue_eval_df['baseline_prediction'].to_numpy(dtype=float) / capacity_values
    rf_rate = revenue_eval_df['rf_prediction'].to_numpy(dtype=float) / capacity_values
    baseline_multiplier = np.where(baseline_rate > 0.85, 1.30, np.where(baseline_rate < 0.50, 0.80, 1.00))
    rf_multiplier = np.where(rf_rate > 0.85, 1.30, np.where(rf_rate < 0.50, 0.80, 1.00))
    perfect_forecast_candidate_demand = actual_values[:, np.newaxis] * price_tiers ** -1.2
    perfect_forecast_candidate_revenue = np.minimum(perfect_forecast_candidate_demand, capacity_values[:, np.newaxis]) * base_price * price_tiers
    perfect_forecast_multiplier = price_tiers[np.argmax(perfect_forecast_candidate_revenue, axis=1)]

    for model_name, selected_multiplier in [
        ('Baseline', baseline_multiplier),
        ('Random Forest', rf_multiplier),
        ('Perfect Forecast', perfect_forecast_multiplier),
    ]:
        demand = actual_values * selected_multiplier ** -1.2
        print(
            f"{model_name} tiers: 1.30x={(selected_multiplier == 1.30).mean():.1%}, "
            f"1.00x={(selected_multiplier == 1.00).mean():.1%}, "
            f"0.80x={(selected_multiplier == 0.80).mean():.1%}; "
            f"demand > capacity={(demand > capacity_values).mean():.1%}"
        )

    elasticity_rows = []
    for elasticity in [-0.5, -1.0, -1.5, -2.0]:
        baseline_revenue = simulate_revenue(
            revenue_eval_df[target_col], revenue_eval_df['baseline_prediction'],
            revenue_eval_df['capacity'], base_price, elasticity=elasticity,
        )
        rf_revenue = simulate_revenue(
            revenue_eval_df[target_col], revenue_eval_df['rf_prediction'],
            revenue_eval_df['capacity'], base_price, elasticity=elasticity,
        )
        perfect_forecast_revenue = calculate_perfect_forecast_revenue(
            revenue_eval_df[target_col], revenue_eval_df['capacity'],
            base_price, elasticity=elasticity,
        )
        baseline_sweep_total = baseline_revenue.sum()
        rf_sweep_total = rf_revenue.sum()
        perfect_forecast_sweep_total = perfect_forecast_revenue.sum()
        elasticity_rows.append({
            'Elasticity': elasticity,
            'Baseline (£)': baseline_sweep_total,
            'RF (£)': rf_sweep_total,
            'Perfect Forecast (£)': perfect_forecast_sweep_total,
            'RF Uplift (%)': (rf_sweep_total - baseline_sweep_total) / baseline_sweep_total * 100
            if baseline_sweep_total else np.nan,
        })
    elasticity_sweep_df = pd.DataFrame(elasticity_rows).set_index('Elasticity')
    print(elasticity_sweep_df.to_string(float_format=lambda value: f"{value:,.2f}"))

    revenue_mismatch = pd.Series({
        'Baseline A': perfect_forecast_total - baseline_total,
        'Baseline B': perfect_forecast_total - interpolation_total,
        'Random Forest': perfect_forecast_total - rf_total,
    })
    fig, ax = plt.subplots(figsize=(12, 6))
    bar_colors = ['tab:blue', 'tab:orange', 'tab:green']
    bars = ax.bar(revenue_mismatch.index, revenue_mismatch.values, color=bar_colors)
    ax.axhline(0, color='tab:red', linewidth=2)
    ax.set_title('Revenue Shortfall vs. Perfect Forecast')
    ax.set_ylabel('Revenue Mismatch (£)')
    ax.yaxis.set_major_formatter(StrMethodFormatter('£{x:,.0f}'))
    ax.grid(True, axis='y', linestyle='--', alpha=0.5)
    ax.legend(handles=[
        Patch(facecolor=bar_colors[0], label='Baseline A: Monthly Carpark Average'),
        Patch(facecolor=bar_colors[1], label='Baseline B: Interpolation Baseline'),
        Patch(facecolor=bar_colors[2], label='Random Forest'),
    ], loc='upper left', bbox_to_anchor=(1.02, 1))
    for bar, value in zip(bars, revenue_mismatch.values):
        ax.annotate(
            f"£{value:,.0f}",
            (bar.get_x() + bar.get_width() / 2, value),
            xytext=(0, 4),
            textcoords='offset points',
            ha='center',
        )
    fig.tight_layout(rect=[0, 0, 0.76, 1])
    plt.show()

    fig, ax = plt.subplots(figsize=(12, 6))
    for column in weekly_revenue_df.columns:
        ax.plot(weekly_revenue_df.index, weekly_revenue_df[column], label=column, linewidth=2)
    ax.set_xlim(all_weeks.min(), all_weeks.max())
    ax.set_xticks(all_weeks[::2])
    ax.set_title('Weekly Revenue by Method')
    ax.set_xlabel('Week of Year')
    ax.set_ylabel('Weekly Revenue (£)')
    ax.yaxis.set_major_formatter(StrMethodFormatter('£{x:,.0f}'))
    ax.grid(True, axis='y', linestyle='--', alpha=0.5)
    ax.legend()
    fig.tight_layout()
    plt.show()

    fig, ax = plt.subplots(figsize=(12, 6))
    for column in [
        'Monthly Carpark Average',
        'Random Forest Model',
        'Perfect Forecast',
    ]:
        ax.plot(cumulative_revenue_df.index, cumulative_revenue_df[column], label=column, linewidth=2)
    ax.set_xlim(all_weeks.min(), all_weeks.max())
    ax.set_xticks(all_weeks[::2])
    ax.set_title('Cumulative Revenue Over Time: Model vs. Baseline vs. Perfect Forecast')
    ax.set_xlabel('Week of Year')
    ax.set_ylabel('Cumulative Revenue (£)')
    ax.yaxis.set_major_formatter(StrMethodFormatter('£{x:,.0f}'))
    ax.grid(True, axis='y', linestyle='--', alpha=0.5)
    ax.legend()
    fig.tight_layout()
    plt.show()

    weekly_perfect_forecast = weekly_revenue_df['Perfect Forecast']
    weekly_revenue_difference_df = pd.DataFrame({
        'Baseline A': weekly_perfect_forecast - weekly_revenue_df['Monthly Carpark Average'],
        'Random Forest': weekly_perfect_forecast - weekly_revenue_df['Random Forest Model'],
        'Baseline B': weekly_perfect_forecast - weekly_revenue_df['Interpolation Baseline'],
    })
    weekly_revenue_difference_pct_df = weekly_revenue_difference_df.div(
        weekly_perfect_forecast.replace(0, np.nan),
        axis=0,
    ) * 100

    fig, ax = plt.subplots(figsize=(12, 6))
    for column in ['Baseline A', 'Random Forest', 'Baseline B']:
        ax.plot(weekly_revenue_difference_df.index, weekly_revenue_difference_df[column], label=column, linewidth=2)
    ax.set_xlim(all_weeks.min(), all_weeks.max())
    ax.set_xticks(all_weeks[::2])
    ax.set_title('Weekly Revenue Shortfall vs. Perfect Forecast')
    ax.set_xlabel('Week of Year')
    ax.set_ylabel('Revenue Difference (£)')
    ax.yaxis.set_major_formatter(StrMethodFormatter('£{x:,.0f}'))
    ax.grid(True, axis='y', linestyle='--', alpha=0.5)
    ax.legend()
    fig.tight_layout()
    plt.show()

    fig, ax = plt.subplots(figsize=(12, 6))
    for column in ['Baseline A', 'Random Forest', 'Baseline B']:
        ax.plot(weekly_revenue_difference_pct_df.index, weekly_revenue_difference_pct_df[column], label=column, linewidth=2)
    ax.set_xlim(all_weeks.min(), all_weeks.max())
    ax.set_xticks(all_weeks[::2])
    ax.set_title('Weekly Revenue Shortfall vs. Perfect Forecast (%)')
    ax.set_xlabel('Week of Year')
    ax.set_ylabel('Revenue Difference (%)')
    ax.yaxis.set_major_formatter(StrMethodFormatter('{x:,.0f}%'))
    ax.grid(True, axis='y', linestyle='--', alpha=0.5)
    ax.legend()
    fig.tight_layout()
    plt.show()

    quarterly_revenue_df = revenue_eval_df.groupby('financial_quarter')[
        ['baseline_revenue', 'interpolation_revenue', 'rf_revenue', 'perfect_forecast_revenue']
    ].sum().reindex([1, 2, 3, 4], fill_value=0)
    quarterly_revenue_df.columns = [
        'Monthly Carpark Average',
        'Interpolation Baseline',
        'Random Forest Model',
        'Perfect Forecast',
    ]
    quarterly_cumulative_revenue_df = quarterly_revenue_df.cumsum()

    fig, axes = plt.subplots(2, 2, figsize=(14, 10), sharey=True)
    for quarter, ax in zip([1, 2, 3, 4], axes.flat):
        quarter_rows = revenue_eval_df[revenue_eval_df['financial_quarter'] == quarter]
        quarter_weekly_revenue_df = quarter_rows.groupby(time_col)[
            ['baseline_revenue', 'interpolation_revenue', 'rf_revenue', 'perfect_forecast_revenue']
        ].sum().sort_index()
        quarter_weekly_revenue_df.columns = quarterly_revenue_df.columns
        quarter_cumulative_weekly_revenue_df = quarter_weekly_revenue_df.cumsum()

        for column in quarter_cumulative_weekly_revenue_df.columns:
            ax.plot(
                quarter_cumulative_weekly_revenue_df.index,
                quarter_cumulative_weekly_revenue_df[column],
                marker='o',
                label=column,
                linewidth=2,
            )
        ax.set_title(f'Financial Quarter {quarter}')
        ax.set_xlabel('Week of Year')
        ax.set_ylabel('Cumulative Revenue (£)')
        ax.yaxis.set_major_formatter(StrMethodFormatter('£{x:,.0f}'))
        ax.grid(True, axis='y', linestyle='--', alpha=0.5)
        ax.legend()
    fig.suptitle('Cumulative Revenue Within Each Financial Quarter')
    fig.tight_layout(rect=[0, 0, 1, 0.96])
    plt.show()
else:
    print('Skipping the Random Forest revenue comparison because run_random_forest=False.')
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

from xgboost import XGBRegressor
from matplotlib.ticker import StrMethodFormatter
from IPython.display import display

# ============================================================
# SETTINGS
# ============================================================
YEAR = 2023
INITIAL_WEEKS = 12
BASE_PRICE = 20.0  # Illustrative £ per occupied space-day

PRICE_MULTIPLIERS = np.linspace(0.70, 1.50, 33)

SCENARIOS = {
    "Roomier capacity": {
        "capacity_factor": 1.25,
        "elasticity": -1.5,
    },
    "Reference": {
        "capacity_factor": 1.00,
        "elasticity": -1.5,
    },
    "Capacity-constrained": {
        "capacity_factor": 0.80,
        "elasticity": -1.5,
    },
}
DISPLAY_SCENARIO = "Capacity-constrained"

KEYS = ["terminal_num", "carpark_name_num"]

BASELINE = "8-week planning baseline"
FOUR_WEEK = "4-week planning baseline"
LAST_WEEK = "Same as last week"
ML = "XGBoost"
ORACLE = "Perfect information"

ALL_METHODS = [BASELINE, FOUR_WEEK, LAST_WEEK, ML]
PLOT_METHODS = [BASELINE, LAST_WEEK, ML]

COLOURS = {
    BASELINE: "#E58B32",
    FOUR_WEEK: "#B79565",
    LAST_WEEK: "#888888",
    ML: "#238B65",
}

# ============================================================
# 1. BUILD THE DATED WEEKLY PANEL
# ============================================================
if not weekly_df.index.is_unique or not weekly_meta.index.is_unique:
    raise ValueError("weekly_df and weekly_meta must have unique indices.")

panel = weekly_df.copy()

# Align metadata to retained rows, accounting for earlier dropna().
panel["date"] = pd.to_datetime(
    weekly_meta.loc[panel.index, "week_start"]
)

if panel["date"].isna().any():
    raise ValueError("Missing week_start dates.")

# Verify the recovered dates agree with the stored week numbers.
date_weeks = panel["date"].dt.isocalendar().week.to_numpy(dtype=int)
stored_weeks = panel["weekofyear"].to_numpy(dtype=int)

if not np.array_equal(date_weeks, stored_weeks):
    raise ValueError("Date metadata does not match weekly_df week numbers.")

panel = panel.loc[panel["date"].dt.year.eq(YEAR)].copy()
panel = panel.sort_values(KEYS + ["date"])

if panel.empty:
    raise ValueError(f"No weekly observations found for {YEAR}.")

if panel[KEYS].isna().any().any():
    raise ValueError("Missing car park or terminal identifiers.")

if panel.duplicated(KEYS + ["date"]).any():
    raise ValueError("Duplicate car park–terminal observations per week.")

if not np.isfinite(panel["occupancy"].to_numpy(float)).all():
    raise ValueError("Occupancy contains missing or non-finite values.")

if panel["occupancy"].lt(0).any():
    raise ValueError("Occupancy must be non-negative.")

panel["unit"] = list(zip(panel[KEYS[0]], panel[KEYS[1]]))

weeks = pd.DatetimeIndex(sorted(panel["date"].unique()))
expected_weeks = pd.date_range(
    weeks.min(), weeks.max(), freq="W-MON"
)

if not weeks.equals(expected_weeks):
    raise ValueError(
        "The source data has missing weeks. Resolve these before "
        "claiming continuous weekly evaluation."
    )

if len(weeks) <= INITIAL_WEEKS:
    raise ValueError("Insufficient history after the initial training period.")

# Keep the same car park–terminal cohort throughout.
counts = panel.groupby("unit")["date"].nunique()
cohort = set(counts.index[counts.eq(len(weeks))])

excluded_units = panel["unit"].nunique() - len(cohort)
panel = panel.loc[panel["unit"].isin(cohort)].copy()

if panel.empty:
    raise ValueError("No car park–terminal series has complete weekly coverage.")

print(
    f"Using {len(cohort)} complete car park–terminal series; "
    f"{excluded_units} incomplete series excluded."
)

# ============================================================
# 2. FEATURES AVAILABLE BEFORE EACH TARGET WEEK
# ============================================================
grouped = panel.groupby(KEYS, sort=False)["occupancy"]

panel["lag_1"] = grouped.shift(1)
panel["lag_2"] = grouped.shift(2)
panel["lag_4"] = grouped.shift(4)

panel["mean_4"] = grouped.transform(
    lambda s: s.shift(1).rolling(4, min_periods=4).mean()
)
panel["mean_8"] = grouped.transform(
    lambda s: s.shift(1).rolling(8, min_periods=8).mean()
)

phase = 2 * np.pi * panel["date"].dt.dayofyear / 365.25
panel["annual_sin"] = np.sin(phase)
panel["annual_cos"] = np.cos(phase)

identity_features = pd.get_dummies(
    panel[KEYS].astype(str),
    prefix=KEYS,
    dtype=float,
)

history_features = [
    "lag_1",
    "lag_2",
    "lag_4",
    "mean_4",
    "mean_8",
    "annual_sin",
    "annual_cos",
]

features = pd.concat(
    [panel[history_features], identity_features],
    axis=1,
)

# Freeze a capacity reference using initial training observations.
# This is a scenario proxy, not measured physical capacity.
initial = panel.loc[panel["date"].isin(weeks[:INITIAL_WEEKS])]
capacity_reference = initial.groupby("unit")["occupancy"].max()

if capacity_reference.le(0).any():
    raise ValueError("Some units have no positive initial capacity reference.")

# ============================================================
# 3. EXPANDING-WINDOW, ONE-WEEK-AHEAD PREDICTIONS
# ============================================================
predictions = []

for target_date in weeks[INITIAL_WEEKS:]:
    train_mask = (
        panel["date"].lt(target_date)
        & features.notna().all(axis=1)
    )
    test_mask = panel["date"].eq(target_date)

    if not train_mask.any():
        raise ValueError(f"No usable training data before {target_date}.")

    if features.loc[test_mask].isna().any().any():
        raise ValueError(f"Missing historical features for {target_date}.")

    # Fresh chronological evaluation with fixed illustrative settings.
    # This does not reuse the old randomly split model.
    model = XGBRegressor(
        objective="reg:squarederror",
        n_estimators=200,
        max_depth=3,
        learning_rate=0.05,
        subsample=0.8,
        colsample_bytree=0.8,
        random_state=42,
        n_jobs=-1,
    )

    model.fit(
        features.loc[train_mask],
        panel.loc[train_mask, "occupancy"],
    )

    batch = panel.loc[
        test_mask,
        ["date", "unit", "occupancy", "mean_4", "mean_8", "lag_1"],
    ].copy()

    batch["baseline_prediction"] = batch["mean_8"]
    batch["four_week_prediction"] = batch["mean_4"]
    batch["last_week_prediction"] = batch["lag_1"]

    batch["model_prediction"] = np.maximum(
        model.predict(features.loc[test_mask]), 0
    )
    batch["capacity_reference"] = batch["unit"].map(capacity_reference)

    predictions.append(batch)

forecast_results = pd.concat(predictions, ignore_index=True)

assert forecast_results.groupby("date").size().eq(len(cohort)).all()
assert forecast_results["date"].nunique() == len(weeks) - INITIAL_WEEKS

print(
    f"Generated forecasts for {forecast_results['date'].nunique()} "
    "consecutive weeks after the initial training period."
)

# ============================================================
# 4. SHARED PRICING OPTIMISER AND DEMAND SIMULATION
# ============================================================
def choose_price(forecast, capacity, elasticity):
    """Select the price expected to maximise revenue under a forecast."""
    forecast = np.maximum(np.asarray(forecast, dtype=float), 0)
    capacity = np.asarray(capacity, dtype=float)

    estimated_demand = (
        forecast[:, None]
        * PRICE_MULTIPLIERS[None, :] ** elasticity
    )
    estimated_revenue = (
        np.minimum(estimated_demand, capacity[:, None])
        * BASE_PRICE
        * PRICE_MULTIPLIERS[None, :]
    )

    return PRICE_MULTIPLIERS[estimated_revenue.argmax(axis=1)]


def realised_revenue(actual, multiplier, capacity, elasticity):
    """Evaluate prices using the same assumed actual demand response."""
    actual = np.asarray(actual, dtype=float)
    demand = actual * multiplier ** elasticity

    return np.minimum(demand, capacity) * BASE_PRICE * multiplier


def evaluate_scenario(frame, capacity_factor, elasticity):
    result = frame.copy()

    capacity = (
        result["capacity_reference"].to_numpy(float)
        * capacity_factor
    )
    actual = result["occupancy"].to_numpy(float)

    # Observed activity is treated as reference-price demand.
    # This is an illustrative assumption, not an estimated demand curve.
    forecast_inputs = {
        BASELINE: result["baseline_prediction"],
        FOUR_WEEK: result["four_week_prediction"],
        LAST_WEEK: result["last_week_prediction"],
        ML: result["model_prediction"],
        ORACLE: actual,
    }

    for name, forecast in forecast_inputs.items():
        multiplier = choose_price(forecast, capacity, elasticity)
        result[f"{name} price"] = multiplier
        result[name] = realised_revenue(
            actual, multiplier, capacity, elasticity
        )

    for name in ALL_METHODS:
        assert np.all(
            result[ORACLE].to_numpy()
            >= result[name].to_numpy() - 1e-7
        ), "Perfect information must provide the scenario revenue ceiling."

    return result


# ============================================================
# 5. REPORT ALL SCENARIOS
# ============================================================
scenario_results = {}
summary_rows = []

for scenario_name, settings in SCENARIOS.items():
    result = evaluate_scenario(forecast_results, **settings)
    scenario_results[scenario_name] = result

    baseline_total = result[BASELINE].sum()
    ml_total = result[ML].sum()
    perfect_total = result[ORACLE].sum()

    available_gap = perfect_total - baseline_total
    uplift = ml_total - baseline_total

    summary_rows.append({
        "Scenario": scenario_name,
        "8-week baseline (£)": baseline_total,
        "4-week baseline (£)": result[FOUR_WEEK].sum(),
        "Last week (£)": result[LAST_WEEK].sum(),
        "XGBoost (£)": ml_total,
        "Perfect information (£)": perfect_total,
        "ML uplift (£)": uplift,
        "ML uplift (%)": (
            100 * uplift / baseline_total
            if baseline_total > 0 else np.nan
        ),
        "Revenue gap recovered (%)": (
            100 * uplift / available_gap
            if available_gap > 1e-6 else np.nan
        ),
        "Different pricing decisions (%)": 100 * (
            result[f"{BASELINE} price"]
            != result[f"{ML} price"]
        ).mean(),
    })

scenario_summary = pd.DataFrame(summary_rows).set_index("Scenario")
display(scenario_summary.round(2))

# ============================================================
# 6. WEEKLY REVENUE AND SHORTFALL
# ============================================================
selected = scenario_results[DISPLAY_SCENARIO]
settings = SCENARIOS[DISPLAY_SCENARIO]

weekly_revenue = (
    selected.groupby("date")[ALL_METHODS + [ORACLE]]
    .sum()
    .sort_index()
)

expected_evaluation = pd.date_range(
    weeks[INITIAL_WEEKS], weeks[-1], freq="W-MON"
)
assert weekly_revenue.index.equals(expected_evaluation)

weekly_shortfall = pd.DataFrame({
    method: weekly_revenue[ORACLE] - weekly_revenue[method]
    for method in ALL_METHODS
})

# Remove only negligible floating-point artefacts.
assert weekly_shortfall.to_numpy().min() >= -1e-6
weekly_shortfall = weekly_shortfall.clip(lower=0)

first_test = weekly_shortfall.index.min()
last_test = weekly_shortfall.index.max()

# ============================================================
# 7. TIMELINE — REMOVE THE EMPTY INITIAL TRAINING AREA
# ============================================================
fig, ax = plt.subplots(figsize=(15, 6.5), facecolor="white")
ax.set_facecolor("white")

quarter_starts = pd.to_datetime([
    f"{YEAR}-01-01",
    f"{YEAR}-04-01",
    f"{YEAR}-07-01",
    f"{YEAR}-10-01",
    f"{YEAR + 1}-01-01",
])

for quarter in range(4):
    start = max(quarter_starts[quarter], first_test)
    end = min(quarter_starts[quarter + 1], last_test)

    if end <= start:
        continue

    if quarter % 2 == 0:
        ax.axvspan(
            start, end,
            color="#EEF2F6", alpha=0.65, zorder=0,
        )

    if first_test < quarter_starts[quarter] < last_test:
        ax.axvline(
            quarter_starts[quarter],
            color="#AAAAAA",
            linestyle="--",
            linewidth=0.8,
        )

    # Avoid a cramped Q1 label for a short end-of-March segment.
    if (end - start).days >= 21:
        ax.text(
            start + (end - start) / 2,
            0.97,
            f"Q{quarter + 1}",
            transform=ax.get_xaxis_transform(),
            ha="center",
            va="top",
            color="#666666",
            fontweight="bold",
        )

for method in PLOT_METHODS:
    ax.plot(
        weekly_shortfall.index,
        weekly_shortfall[method],
        label=method,
        color=COLOURS[method],
        linestyle=":" if method == LAST_WEEK else "-",
        linewidth=2.5,
        marker="o",
        markersize=3,
        zorder=3,
    )

ax.set_xlim(first_test, last_test)
maximum = weekly_shortfall[PLOT_METHODS].to_numpy().max()
ax.set_ylim(0, max(maximum * 1.22, 1))

ax.xaxis.set_major_locator(mdates.MonthLocator())
ax.xaxis.set_major_formatter(mdates.DateFormatter("%b"))
ax.yaxis.set_major_formatter(StrMethodFormatter("£{x:,.0f}"))
ax.set_ylabel("Simulated weekly revenue shortfall (£)", fontsize=12)

fig.suptitle(
    "How much revenue opportunity does each forecast miss?",
    x=0.08, y=0.98,
    ha="left",
    fontsize=18,
    fontweight="bold",
)
fig.text(
    0.08, 0.915,
    f"{DISPLAY_SCENARIO} scenario · Capacity set to "
    f"{settings['capacity_factor']:.0%} of each car park’s initial "
    "observed peak weekly activity\n"
    f"Assumed price elasticity: {settings['elasticity']:.1f} · "
    "Same pricing optimiser for every method · Lower is better",
    fontsize=10.5,
    color="#555555",
    va="top",
)

ax.grid(axis="y", alpha=0.15)
ax.set_axisbelow(True)
ax.legend(
    loc="upper center",
    bbox_to_anchor=(0.5, -0.12),
    ncol=3,
    frameon=False,
)

for spine in ["top", "right"]:
    ax.spines[spine].set_visible(False)

fig.text(
    0.08, 0.055,
    f"Evaluation begins {first_test:%d %b %Y}: the first "
    f"{INITIAL_WEEKS} weeks establish training history. "
    "Earlier weeks are omitted, not zero revenue.",
    fontsize=9,
    color="#555555",
)
fig.text(
    0.08, 0.025,
    "Illustrative simulation, not measured financial impact. "
    "Capacity is a scenario proxy. Q2–Q4 denote calendar quarters.",
    fontsize=9,
    color="#666666",
)
fig.subplots_adjust(
    left=0.08, right=0.98, top=0.79, bottom=0.23
)
plt.show()

# ============================================================
# 8. FINAL BAR CHART — ALL THREE METHODS
# ============================================================
losses = weekly_shortfall[PLOT_METHODS].sum()

fig, ax = plt.subplots(figsize=(10, 4.8), facecolor="white")

bars = ax.barh(
    PLOT_METHODS,
    losses.reindex(PLOT_METHODS).to_numpy(),
    color=[COLOURS[method] for method in PLOT_METHODS],
    height=0.55,
)

ax.invert_yaxis()
ax.bar_label(
    bars,
    labels=[f"£{losses[method]:,.0f}" for method in PLOT_METHODS],
    padding=8,
    fontsize=12,
    fontweight="bold",
)

ax.set_xlim(0, max(losses.max(), 1) * 1.3)
ax.xaxis.set_major_formatter(StrMethodFormatter("£{x:,.0f}"))
ax.set_xlabel("Total simulated revenue shortfall (£) · Lower is better")
ax.set_title(
    "Which forecast gets closest to perfect information?",
    loc="left",
    fontsize=16,
    fontweight="bold",
    pad=22,
)
ax.grid(axis="x", alpha=0.12)
ax.set_axisbelow(True)

for spine in ["top", "right"]:
    ax.spines[spine].set_visible(False)

fig.text(
    0.02, 0.025,
    f"{DISPLAY_SCENARIO} scenario · "
    f"{first_test:%d %b}–{last_test:%d %b %Y} · "
    "Same evaluated weeks, capacity assumptions and pricing choices",
    fontsize=9,
    color="#666666",
)
fig.tight_layout(rect=[0, 0.08, 1, 1])
plt.show()

# ============================================================
# 9. PRESENTATION SENTENCES — BASED ON THE ACTUAL RESULTS
# ============================================================
ml_revenue = weekly_revenue[ML].sum()

print("\nSIMULATED PERFORMANCE COMPARISONS")

for comparator in [BASELINE, LAST_WEEK, FOUR_WEEK]:
    comparator_revenue = weekly_revenue[comparator].sum()
    difference = ml_revenue - comparator_revenue

    if abs(difference) < 1e-6:
        print(f"XGBoost matched {comparator}'s simulated revenue.")
    else:
        direction = "more" if difference > 0 else "less"
        percentage = (
            100 * abs(difference) / comparator_revenue
            if comparator_revenue > 0 else np.nan
        )
        percentage_text = (
            f" ({percentage:.2f}%)"
            if np.isfinite(percentage) else ""
        )

        print(
            f"XGBoost generated £{abs(difference):,.0f}"
            f"{percentage_text} {direction} simulated revenue "
            f"than {comparator}."
        )

baseline_loss = losses[BASELINE]
gain = ml_revenue - weekly_revenue[BASELINE].sum()

if baseline_loss > 1e-6:
    print(
        f"Revenue gap recovered versus the eight-week baseline: "
        f"{100 * gain / baseline_loss:.1f}%."
    )
else:
    print(
        "The eight-week baseline has no material gap to perfect information; "
        "a gap-recovery percentage is not meaningful."
    )
```
